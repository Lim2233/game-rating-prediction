```python
import pandas as pd

# 读取原始数据
df = pd.read_csv('../data/raw/games_raw.csv')

cols_to_drop = ['img', 'last_update', 'title','na_sales','jp_sales','pal_sales','other_sales']
df.drop(columns=cols_to_drop, inplace=True)

df.dropna(subset=['critic_score'],inplace = True)

for i in df.columns:
        
    df.dropna(subset=[i],inplace = True)
    

save_road='../data/processed/data_processed.csv'
df.to_csv(save_road,index=False)

df.info()
```

```python

df = pd.read_csv('../data/processed/data_processed.csv')

# 先转换为 datetime 类型，指定格式（dayfirst=True 表示日在前）
df['release_date'] = pd.to_datetime(df['release_date'], format='%d-%m-%Y', errors='coerce')

df.head()
```

```python
df['release_date'].max()
```

```python
# 获取当前日期（仅日期部分，忽略时间）
ddl = pd.Timestamp('2024-01-01')
ddl
```

```python

# 计算距今天数（单位：天）
df['days_since_release'] = (ddl - df['release_date']).dt.days

df.head()
```

```python
df.drop(columns='release_date',inplace=True)
df.head()
```

```python
# 对多个列
cat_cols = ['console', 'genre', 'publisher', 'developer']
for col in cat_cols:
    df[f'{col}_code'] = df[col].astype('category').cat.codes
df.head()
```

```python
df.info()
```

```python
df.dropna(subset=['days_since_release'],inplace = True)
```

```python
df.dropna(subset=['total_sales'],inplace = True)
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

```python
df.drop(columns=['console_code','genre_code','publisher_code','developer_code'],inplace=True)
df.info()
```

```python
df.describe(include='all')
```

```python
from catboost import CatBoostRegressor, Pool
```

```python
# 定义特征列和目标列
features = ['console', 'genre', 'publisher', 'developer', 'total_sales', 'days_since_release']
target = 'critic_score'

X = df[features]
y = df[target]

print(type(X))
```

```python
X
```

```python
# 获取所有字符串类型的列名
categorical_features = X.select_dtypes(include=['object', 'string']).columns.tolist()
```

```python
categorical_features
```

```python
import numpy as np
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.05, random_state=42
)
```

```python
X_train.info()
```

```python
model = CatBoostRegressor(
    iterations=1500,           # 树的数量（迭代次数）
    learning_rate=0.05,       # 学习率
    depth=8,                  # 树的深度
    loss_function='RMSE',     # 回归任务常用 RMSE
    cat_features=categorical_features,  # 传入类别特征
    verbose=100,              # 每100轮输出一次日志
    early_stopping_rounds=100  # 早停，防止过拟合
)

# 训练模型（使用 Pool 不是必须的，但可以更方便地传递参数）
model.fit(X_train, y_train, eval_set=(X_test, y_test))
```

```python
y_pred = model.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)

print(f"RMSE: {rmse:.4f}")
print(f"MAE: {mae:.4f}")
```
